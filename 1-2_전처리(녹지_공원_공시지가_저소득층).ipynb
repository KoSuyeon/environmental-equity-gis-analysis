# 공간데이터

- 공공공지: 사유지가 아닌 국공유지로 향후에 발생할 수 있는 공공용도의 토지이용에 대응할 수 있다 -> 제외

```python
import pandas as pd
pd.options.display.float_format = '{:.8f}'.format

gis_data = pd.read_csv('data/한국토지주택공사_공간시설 현황정보_20220802.csv',encoding='cp949')
gis_data = gis_data[gis_data['시도']=='서울특별시']


# 공원
gongwon = gis_data[gis_data['중분류'].isin(['(기존)도시자연공원','근린공원','어린이공원','소공원','역사공원','수변공원','묘지공원','체육공원','문화공원'])]
gongwon.drop_duplicates(keep='first', inplace=True)
final_gongwon = pd.DataFrame(gongwon.groupby(['시군구','읍면동']).agg({'읍면동':"count", '결정면적':"mean"}).rename(columns={'읍면동': '공원_개수', '결정면적': '공원_면적'})).reset_index()

# 녹지
nokji = gis_data[gis_data['중분류'].isin(['완충녹지','경관녹지','연결녹지'])]
nokji.drop_duplicates(keep='first', inplace=True)
final_nokji = pd.DataFrame(nokji.groupby(['시군구','읍면동'], group_keys=False).agg({'읍면동':"count", '결정면적':"mean"}).rename(columns={'읍면동': '녹지_개수', '결정면적': '녹지_면적'})).reset_index()
```

### 법정동코드

```python
import pandas as pd
법정동코드 = pd.read_excel('data/한국행정구역분류_정제.xlsx',sheet_name='법정동')
```

```python
법정동코드
```

```python
# 데이터프레임 합치기
# 공원
final_gis = pd.merge(법정동코드,final_gongwon, left_on=['시군구','법정동'],right_on=['시군구','읍면동'], how='left')
final_gis.drop(columns='읍면동', inplace = True)
# 녹지
final_gis = pd.merge(final_gis,final_nokji, left_on=['시군구','법정동'],right_on=['시군구','읍면동'], how='left')
final_gis.drop(columns='읍면동', inplace = True)

# nan값 채우기
final_gis.fillna(0, inplace=True)
```

# 공시지가 - 데이터 구해서 추후 다시 하기로 함

```python
df = df[['YEAR','BASE_MON','JIGA','BJDONG_CD']]
df_jiga=df.groupby(['YEAR','BASE_MON','BJDONG_CD']).mean()
```

# 저소득층

```python
저소득층 = pd.read_csv('data/SE_FN_CRD_STAT_H_DONG.csv',encoding='cp949')
저소득층=저소득층[['DO_NM','CT_NM','H_DNG_NM','total']]
저소득층_비율 = pd.DataFrame(저소득층.groupby(['CT_NM','H_DNG_NM']).agg({'total':"mean"})).reset_index().rename(columns={'CT_NM':'시군구','H_DNG_NM':'행정동명','total': '저소득층_비율'})                                                                                                         
```

```python
저소득층_비율
```

```python
행정동코드 = pd.read_excel('data/한국행정구역분류_정제.xlsx',sheet_name='행정동')
df_행정동코드=행정동코드[['시군구','행정구역명','행정기관코드']].rename(columns={'행정구역명':'행정동명'})
final_저소득층=pd.merge(df_행정동코드,저소득층_비율)

final_저소득층.to_csv('전체_중_저소득층_비율.csv',encoding='cp949',index=False)
```
