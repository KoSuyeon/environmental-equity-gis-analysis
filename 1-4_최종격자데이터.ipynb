# 버스, 지하철

```python
import pandas as pd
df = pd.read_csv('최종데이터/버스,지하철 결합 격자.csv',encoding='cp949')
df.drop('left',axis=1,inplace=True)
df.drop('top',axis=1,inplace=True)
df.drop('right',axis=1,inplace=True)
df.drop('bottom',axis=1,inplace=True)
```

# 재생사업

```python
재생 = pd.read_excel('최종데이터/도시재생_개수.xlsx').rename(columns={'동분류위_3':'행정동코드'})
df_재생 = 재생.groupby('행정동코드').sum().reset_index()

df = pd.merge(df,df_재생,how='left',on='행정동코드').rename(columns={'도시재생사':'도시재생사업수'})
df['도시재생사업수'].fillna(0,inplace=True)
df.drop('id',axis=1,inplace=True)
```

# 소득 및 인구

```python
a = pd.read_csv('최종데이터/행정동_기초수급자인구밀도.csv',encoding='cp949')[['시군구','행정동','인구밀도 (인원 수 / 면적 (k㎡))']]
b = pd.read_csv('최종데이터/행정동_저소득가구_비율.csv',encoding='cp949')
c = pd.read_csv('최종데이터/행정동_평균생활인구수(야간).csv',encoding='cp949')[['시군구','행정동','행정동코드','행정동별_평균생활인구수']].rename(columns={'행정동별_평균생활인구수':'야간생활인구'})
d = pd.read_csv('최종데이터/행정동_평균생활인구수(주간).csv',encoding='cp949')[['시군구','행정동','행정동코드','행정동별_평균생활인구수']].rename(columns={'행정동별_평균생활인구수':'주간생활인구'})
e = pd.read_csv('최종데이터/행정동_평균생활인구수(주야간).csv',encoding='cp949')[['시군구','행정동명','행정동코드','행정동별_평균생활인구수_합계']].rename(columns={'행정동명':'행정동','행정동별_평균생활인구수_합계':'총생활인구'})
                
a1 = pd.merge(a,b)
c1 = pd.merge(c,d)
e1 = pd.merge(c1,e,on='행정동코드')

df_소득인구 = pd.merge(a1,e1,on='행정동코드',how='outer')[['행정동코드','인구밀도 (인원 수 / 면적 (k㎡))','저소득층_비율','주간생활인구','야간생활인구','총생활인구']]
```

```python
df = pd.merge(df,df_소득인구,how='left',on='행정동코드')
```

```python
df
```

# 기온

```python
import pandas as pd
df = pd.read_csv('최종데이터.csv',encoding='cp949')
```

```python
data=pd.read_csv('data/기온데이터.csv',encoding='cp949')
data['시군구'] = data['지점명'].map(lambda x:x+'구')
```

```python
final=data.groupby(['시군구']).agg({"평균기온(°C)":'mean'}).reset_index()
final.to_csv('구별기온.csv',encoding='cp949',index=False)
```

```python
a = pd.merge(df,final, how='left')
b = pd.merge(df,a,how='left')
b.to_csv('최종데이터.csv',encoding='cp949',index=False)
```
