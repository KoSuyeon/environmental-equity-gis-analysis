```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Amount에 log변환을 취해서 분포도 변환시킨후 학습시키기
def get_preprocessed_df(df):
    df_copy = df.copy()
    amount_n = np.log1p(df['기초수급자_인구밀도'])
    df_copy.insert(0, 'ln_기초수급자_인구밀도', amount_n)
    df_copy.drop('기초수급자_인구밀도', axis=1, inplace=True)
    return df_copy
    
ln_df = get_preprocessed_df(df)

plt.figure(figsize=(8,4))
plt.xticks(range(0, 30000, 1000), rotation=60)
sns.distplot(ln_df['ln_기초수급자_인구밀도'])
```
