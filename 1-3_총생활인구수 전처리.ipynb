# 위경도 추가하기

```python
from geopy.geocoders import Nominatim
geo_local = Nominatim(user_agent='South Korea')
```

```python
def geocoding(address):
    try:
        geo = geo_local.geocode(address)
        x_y = [geo.latitude, geo.longitude]
        return x_y

    except:
        return [0,0]
```

```python
latitude = []
longitude =[]

for i in mean_주간['주소']:
    latitude.append(geocoding(i)[0])
    longitude.append(geocoding(i)[1])
```

```python
address_주간 = pd.DataFrame({'상세주소':mean_주간['주소'],'위도':latitude,'경도':longitude, '행정동별_평균생활인구수':mean_주간['행정동별_평균생활인구수']})
```

```python
address_주간
```

```python
address_주간.to_csv('주소_행정동별_평균생활인구수(주간).csv')
```
