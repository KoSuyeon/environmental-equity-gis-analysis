```python
import pandas as pd
df = pd.read_csv('data/한국토지주택공사_공간시설 현황정보_20220802.csv',encoding='cp949')
```

```python
df.head(3)
```

```python
df_공원녹지 = df[(df.시도=='서울특별시')&(df.대분류.isin(['공원','녹지']))].groupby('시군구').agg({'결정면적':'sum'}).reset_index()
```

```python
final_공원녹지 = df_공원녹지[df_공원녹지.시군구 != '서울특별시'].reset_index(drop=True)
```

```python
final_공원녹지.to_csv('구별_공원녹지면적.csv',encoding='cp949',index=False)
```
